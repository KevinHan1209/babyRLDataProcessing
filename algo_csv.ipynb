```python
import pandas as pd
```

```python
df = pd.read_csv('merged_rnd_multi_seed.csv')
```

```python
# Get unique object types that have 'attached' state
attached_objects = df[df['state_name'] == 'kicked']['object_type'].unique()
print("Objects that can be kicked:", attached_objects)
# Get activity counts for kicked state by object type
kicked_counts = df[df['state_name'] == 'kicked'].groupby('object_type')['activity_count'].sum()
print("\nKicked state activity counts by object:")
print(kicked_counts)

```
